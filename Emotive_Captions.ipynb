```python
import google.generativeai as genai
from google.colab import files

# 1. Setup
genai.configure(api_key="AIzaSyBhtMqKLoayuHqvtWvQh4OQ7rHzmD7q1xI")
model = genai.GenerativeModel('gemini-3-flash-preview')

# 2. Upload your audio file
print("Upload the audio you want to analyze for tone:")
uploaded = files.upload()
audio_file_path = list(uploaded.keys())[0]

# 3. Upload to Gemini Files API
audio_file = genai.upload_file(path=audio_file_path)

# 4. The "Prosody Prompt"
# This is the 'secret sauce' that tells the AI to look for the music, not just the words.
prompt = """
Transcribe this audio into a caption format.
Crucially, I want you to 'code' the prosody.
Insert labels in brackets like [jokingly], [sarcastically], [hesitantly], or [with rising excitement]
whenever the speaker's tone, rhythm, or volume changes the meaning of the words.
Format it as a timestamped transcript.
"""

# 5. Generate
print("AI is listening for 'hidden signals' in the speech...")
response = model.generate_content([prompt, audio_file])

print("\n--- EMOTIVE TRANSCRIPT ---\n")
print(response.text)

with open("emotive_transcript.txt", "w") as f:
    f.write(response.text)
```
